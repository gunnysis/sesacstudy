```python
# https://www.eundunmaeul.store/
# 해당 사이트 웹크롤링
# 1. requests 라이브러리를 이용하여 웹페이지의 HTML 소스를 가져온다.
# 2. BeautifulSoup 라이브러리를 이용하여 HTML 소스에서 제목들을 가져온다.
import requests
from bs4 import BeautifulSoup
url = 'https://www.eundunmaeul.store/'
response = requests.get(url)
soup = BeautifulSoup(response.text, 'html.parser')
titles = soup.find_all('h2', class_='entry-title')
for title in titles:
    print(title.text.strip())
    



```
