```python
import datetime

dir(datetime)
```

```python
my_date = datetime.datetime(2024, 6, 1,11,52)

weeknames= ['월', '화', '수', '목', '금', '토', '일']
print(weeknames[my_date.weekday()])
now = datetime.datetime.now()
yyyy = now.strftime('%Y')
mm = now.strftime('%m')
dd = now.strftime('%d')
print(f'{yyyy}년 {mm}월 {dd}일')

```

```python
first_date = datetime.datetime(2025, 1, 1)
list_date = [(first_date + datetime.timedelta(days=i)).strftime('%Y-%m-%d') for i in range(60)]
print(list_date)
```

```python
from dateutil.relativedelta import relativedelta
help(relativedelta)
```

```python
date1 = datetime.datetime(2024, 6, 1)
date2 = date1 + relativedelta(months=2)

print(date2.strftime('%Y-%m-%d'))
```

```python
new_date = date1 - relativedelta(months=2)
print(new_date.strftime('%Y-%m-%d'))
```
